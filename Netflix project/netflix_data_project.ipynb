## Data Visualization

```python
netflix.groupby("Region")["Monthly_Revenue"].sum().plot(kind="bar", ylabel="Monthly Revenue", title="Region Wise Revenue", rot=0, edgecolor="black")
```

```python
netflix.groupby("Subscription_Plan")["Rating"].sum().plot(kind="pie", title="Subscription Plan Wise Rating", ylabel="", autopct="%1.1f%%")
```

```python
netflix["Rating"].value_counts().plot(kind="bar", title="Distribution of Ratings", xlabel="Rating", ylabel="Count", rot=0, color="green", edgecolor="black")
```

```python
netflix.groupby("Category")["Monthly_Revenue"].sum().plot(kind="pie",title="Category Wise Revenue", ylabel="", autopct="%1.1f%%")
```

```python
netflix.groupby("Watch_Date")["Monthly_Revenue"].sum().plot(kind="line", title="Revenue Trend Over Time", xlabel="Watch Date", ylabel="Total Revenue", marker="o", color="green")
```

```python
netflix.groupby("Subscription_Plan")["Monthly_Revenue"].sum().plot(kind="bar", title="Revenue by Subscription Plan", xlabel="Subscription Plan", ylabel="Total Revenue", color="purple", edgecolor="black", rot=0)
```

```python
netflix.groupby("Device")["Customer_ID"].count().plot(kind="pie", title="Most Used Devices for Netflix", ylabel="", autopct="%1.1f%%")

```
